````python
l = ['this', False, 56.55, ['t', 'h', 'i', 's']]
for ele in enumerate(l):
    print(f"{ele =}")
````

````python
l = ['this', False, 56.55, ['t', 'h', 'i', 's']]
for ele in enumerate(l):
    idx , e= ele
    print(f"{idx =}, {e =}")
````

````python
l = ['this', False, 56.55, ['t', 'h', 'i', 's']]
for idx, e in enumerate(l):
    print(f"{idx =}, {e =}")
````

````python
d = { 'a':1, 'b':2}
for ele in d:
    print(f"{ele = }")
````

````python
d = { 'a':1, 'b':2}
for key in d:
    val = d[key]
    print(f"{key = },{val = }")
````

````python
for key in d.items():
    print(f"{key = }")
````

````python
for k,v in d.items():
    print(f"{k = },{v = }")
````

# While loop
```python
#syntax
while condition:
    #statements
```

````python
# Wtite a program to take input from the user and print it until user inputs zero(0).
user_input = input("Enter the value: ")
if user_input.isnumeric():
    user_input = int(user_input)

while user_input != 0:
    print(f"{user_input = }")

    user_input = input("Enter the value: ")
    if user_input.isnumeric():
        user_input = int(user_input)
    else:
        user_input = 0
````

````python
# Wtite a program to take input from the user and print it until user inputs zero(0).
user_input = input("Enter the value: ")
if user_input.isnumeric():
    user_input = int(user_input)

while user_input != 0:
    print(f"{user_input = }")

    for i in range(3):
        user_input = input("Enter the value: ")
        if user_input.isnumeric():
            user_input = int(user_input)
            break
        if i != (3-1):
            print("Retrying...")
    if not isinstance(user_input, int):
        break
````
