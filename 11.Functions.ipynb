### Positional Arguments in function call

````python
def add_values(a, b):
    total = a+b
    return total
````

````python
add_values(1, 2)
````

````python
add_values()
````

### keyword-based Arguments in function call

````python
def add_values(a, b):
    total = a+b
    return total

add_values(a = 5, b = 6)
````

````python
add_values(b = 6, a = 5)
````

### both positional and keyword-based arguments in function call

````python
add_values(5, b = 6)
````

````python
add_values(a = 5, 6) #positional argument follows keyword argument
````

### Default parameters

````python
def add_values(a, b, c=0):
    total = a + b + c
    return total

add_values(1, 2)
````

````python
add_values(1, 2, 3)
````

````python
add_values(1, b=2, c=3)
````

````python
add_values(1, 2, c=3)
````

## `Args` and `Kwargs`

````python
def add_values(*args):
    print(f"{type(args)= }")
    print(f"{args= }")

add_values()
````

````python
add_values(1)
````

````python
add_values(100, 200, 300, 400)
````

````python
def add_values(**kwargs):
    print(f"{type(kwargs)= }")
    print(f"{kwargs= }")

add_values()
````

````python
add_values(a=1, b=2, c=3, d=4)
````

````python
def add_values(*args, **kwargs):
    print(f"{type(args)= }")
    print(f"{args= }")
    print(f"{type(kwargs)= }")
    print(f"{kwargs= }")

add_values()
````

### Anonymous function/ lambda function

````python
def add (a, b):
    return a + b
add(2,5)
````

````python
total = lambda a, b: a + b
total(2,5)
````

````python
def square(ele):
    return ele ** 2

sq = lambda ele: ele ** 2
````

````python
l =[ -0.5, 0.25, 1, -2, 3]
l.sort()
l
````

````python
l =[ -0.5, 0.25, 1, -2, 3]
l.sort(key = square)
l
````

````python
l =[ -0.5, 0.25, 1, -2, 3]
l.sort(key = lambda ele: ele ** 2)
l
````
