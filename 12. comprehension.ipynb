```python
# write a rpogram to transform the elements in the list into new list using following condition.
# multiply by 2 if its even ignore otherwise
l = [1, 2, 3, 4, 5]


for _ in range(4):
    if l[_] % 2 == 0:
        l[_]= l[_]*2
print (l)
```

```python
# list comprehension

l = [1, 2, 3, 4, 5]
[ele * 2 for ele in l if ele % 2 == 0]
```

```python
l = [1, 2, 3, 4, 5]
result = []

# loop through each element
for ele in l:
    if ele % 2 == 0:   # check if even
        result.append(ele * 2)  # multiply by 2 and add to new list
print("Transformed list:", result)
```

```python
l = [1, 2, 3, 4, 5]
result = []

# loop through each element
for ele in l:
    if ele % 2 == 0:   # check if even
        result.append(ele * 2) 
    else:
        result.append(ele * 100)
print("Transformed list:", result)
```

```python
l = [1, 2, 3, 4, 5]
[(ele * 2 if ele % 2 == 0 else ele * 100) for ele in l]
```

```python
l = [1, 2, 3, 4, 5]
result = []

# loop through each element
for ele in l:
    if ele > 2:
        if ele % 2 == 0:   # check if even
            result.append(ele * 2) 
        else:
            result.append(ele * 100)
result
```

```python
d = {'a':1 ,'b':2, 'c':3, 'd':4, 'e':5}
result={}
for k,v in d.items():
    if v % 2 == 0:
        result[k] = v * 100
    else:
        result[k] = v + 100
result
```
