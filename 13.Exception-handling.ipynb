### Exception Handling
```python
try:
    # try block statement
    # statements with possibility of error
except:
    # action to take if error occurs.
finally:
    # runs everytime.
    # optional
```

````python
value = input("Enter numeric value: ")
value = int(value)

try:
    result = 100 / value
except Exception as e:
    print(f'Error occured: {e}')
    result = -1
finally:
    print(f"{result = }")
````

````python
value = input("Enter numeric value: ")
value = int(value)

try:
    result = 100 / value
except ZeroDivisionError as e:
    print(f'Error occured: {e}')
    result = -1
finally:
    print(f"{result = }")
````

````python
value = input("Enter numeric value: ")
value = int(value)

try:
    result = 100 / value
except NameError as e:
    print(f'Error occured: {e}')
    result = -1
finally:
    print(f"{result = }")
````

````python
d = {'a':1, 'b':2, 'c':3}

user_key = input("Enter the key: ")

try:
    value = d[user_key]
    print(value)
except (KeyError, NameError) as e:
    print(f"Error: {e}")
````
