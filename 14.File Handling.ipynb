# File Handling
- uses `open()` function with different modes.
- commonly used modes are read(r), write(w), append(a)

## Write (w) mode
- creates new file if file doesn't exists at specified path.
- if file already exists at specific path, then all the contents are erased to make the file as new one.

```python
abs_file_path = "C:/Shankar Adhikari/python/files/using_abs.txt"
rel_fiel_path2 = "files/using_rel.txt"
```

```python
f = open(abs_file_path, 'w')
```

```python
f.write("This statement is written from \nwrite method.")
```

```python
f.close()
```

```python
f = open(abs_file_path, 'w')
```

```python
l = ["This is car.", "This is nice."]
f.writelines(l)
```

```python
f.close()
```

## Append ('a') Mode
- Creates new file if the file is not present in the specified path as like write .
- If the file exists in the specified path, no change is made in the content of file.
- Write and sriteline method can be used when opened in append mode.
- Adds new content to the end of the existing content.

```python
rel_file_path = "files/using_abs.txt"
f = open(rel_file_path, 'a')
```

```python
f.write('\nThis is appended text.')
f.close()
```

```python
f = open(rel_fiel_path2, 'a')
f.close()
```
