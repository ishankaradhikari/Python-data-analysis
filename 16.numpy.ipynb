### 3-D Array

```python
l = [[
    [1, 2],
    [3, 4]
    ],
    [
        [5, 6],
        [7, 8]
    ]
]
arr3 = numpy.array(l)
print(arr3)
print(type(arr3))
print(arr3.ndim)
print(arr3.shape)
```

```python
arr3[1,0,1]
```
