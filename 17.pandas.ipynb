## Creating series

```python
import pandas as pd
```

```python
l = [1, 2, 3]
s1 = pd.Series(l)
print(s1)
print(type(s1))
```

```python
l = (1, 2, 3)
s1 = pd.Series(l)
print(s1)
print(type(s1))
```

```python
import numpy as np
arr = ([1, 2, 3])
s1 = pd.Series(arr)
print(s1)
print(type(s1))
```

```python
l = [1, 2, 3]
s1 = pd.Series(l, name = "Demo")
print(s1)
print(type(s1))
```

```python
l = [1, 2, 3]
labels = ['first', 'second', 'third']
s1 = pd.Series(l, name = "Demo", index = labels)
print(s1)
print(type(s1))
```
