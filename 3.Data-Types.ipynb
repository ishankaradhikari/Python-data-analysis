- % operator for string formatting

```python
'this is a %s' %'class'
```

```python
'this is a %d' %1
```

- format() method fro string fromatting.

```python
a = "this is a {}"
a
```

```python
a.format('class')
```

```python
"this {} a {}".format('is','class')
```

```python
"this {1} a {0}".format('class', 'is')
```

```python
"this {verb} a {name}".format(verb = 'is', name = 'class')
```

```python
c = 'is'
d = 'class'

"this {verb} a {name}".format(verb=c , name=d )
```

- f-string string formatting

```python
c = 'is'
d = 'class'
f"this {c} is {d}"
```
