# Tuples
- tuples are immutable (elements in tuple neither can be removed nor can be added)

```python
type(4)
```

```python
# empty tuple
t1 = ()
type(t1)
```

```python
t2=(1,2,3)
type(t2)
```

```python
t3=(1,)
type(t3)
```

```python
t4=(1, 2, 3, 4, 5, 2, 3, 2)
```

```python
t4.count(3)
```

```python
t4.index(4)
```

```python
t4.index(2)
```

```python
t4.index(2, 6) #(element of which the index is to be searched, from which index to start)
```
