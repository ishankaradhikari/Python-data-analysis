# Dictionary
- uses curly braces to define the structure
- stores data as key-value pairs.
- key must be immutable type and value has no lilmitation in data types.
```python
# Syntax
d = {} # empty dictionary
d = dict() #empty dictionary
d = {'key1' : 'value1' , 'key2' : 'value2'} #default key-value pairs
d = dict(key1 = value , key2 = value2)
```

````python
d = {}
print(d)
print(type(d))
````

````python
d = dict()
print(d)
print(type(d))
````

````python
d = {'k1':'v1' , 'k2':'v2'}
print(d)
print(type(d))
````

````python
d = dict(k1='v1', k2='v2')
print(d)
print(type(d))
````

````python
# list can't be used as key in dictionary as it is mutable implying unhashable
# l = [1, 2]
# d1 = {l : 'val'}
# print(d1)
````

````python
# tuple can be used as key in dictionary as it is immutable implying hashable 
l = (1, 2)
d1 = {l : 'val'}
print(d1)
````
