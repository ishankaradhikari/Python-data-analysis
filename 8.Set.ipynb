# Set
- Stores `unique values` only.
- Use curly braces similar to dictionary but stores elements only not the key-value pair
- Mathematical set operations can be performed
- `Indexing is not possible` as elements are stored in out of order
- mutable

```python
# 
s1= set()
print(s1)
print(type(s1))
```

```python
s1.add(1)
s1
```

```python
s1.add(2)
s1.add(3)
s1.add(21)
s1.add(223)
s1.add(6)
s1.add(32)
s1
```

```python
s1.remove(223)
s1
```

```python
s1.remove(5000)
```

```python
s1.discard(5000)
```

```python
return_value = s1.pop()
print(f"{return_value =}")
print(f"{s1= }")
```

```python
s1 = {1, 2, 3, 4, 5, 6}
s2 = {2, 4, 7}

s1.union(s2)
```

```python
s1 = {1, 2, 3, 4, 5, 6}
s2 = {2, 4, 7}
s1.intersection(s2)
```

```python
s1 = {1, 2, 3, 4, 5, 6}
s2 = {2, 4, 7}
s1.intersection_update(s2)

print(f"{s1 = }")
print(f"{s2 =}")
```

```python
s1 = {1, 2, 3, 4, 5, 6}
s2 = {2, 4, 7}
diff= s1.difference(s2)

print(f"{s1 = }")
print(f"{s2 =}")
print(f"{diff = }")
```
