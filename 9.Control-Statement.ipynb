# SWITCH
- `match` keyword is used .
```python
x = 1
exp = x**2 + x - 1
match exp:
    case 1:
        print(f"for {x = }")
    case 5:
        print(f"for{x = }")
    case 11:
        print(f"for{x = }")
    case _:
        print(f"default case")
```

````python
x = 10
exp = x**2 + x - 1
match exp:
    case 1:
        print(f"for x = 1")
    case 5:
        print(f"for x = 2")
    case 11:
        print(f"for x = 3")
    case _:
        print(f"default case")
````

# Nested if-else

````python
value = 4
if value < 4:
    print("lt 4")
else:
    if value == 4:
        print("eq to 4")
    else:
        print("gt 4")
````
